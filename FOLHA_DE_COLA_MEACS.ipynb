## 5. DETECTAR OUTLIERS — regra 1.5 x AIQ
Usando df_censo, coluna renda_per_capita

```python
# QUEM SAO OS OUTLIERS?
col = 'renda_per_capita'
Q1 = df_censo[col].quantile(0.25)
Q3 = df_censo[col].quantile(0.75)
AIQ = Q3 - Q1
lim_inf = Q1 - 1.5 * AIQ
lim_sup = Q3 + 1.5 * AIQ

outliers = df_censo[(df_censo[col] < lim_inf) | (df_censo[col] > lim_sup)]
print(f"Limite inferior: R${lim_inf:.2f}")
print(f"Limite superior: R${lim_sup:.2f}")
print(f"Outliers encontrados: {len(outliers)}")
print()
print(outliers[['estado', col]])
```

**INTERPRETACAO:** "Pela regra 1.5xAIQ, o Distrito Federal (R$1715) ultrapassa o limite superior e e considerado um outlier. Por ser a capital federal, concentra funcionalismo publico com salarios mais altos, o que explica a renda atipicamente elevada em comparacao aos demais estados."

## 9. RESPOSTAS TEORICAS PRONTAS — adapte so os valores

### Media > Mediana → o que isso significa?

**Resposta pronta (so troca os numeros):**
"O fato de a media (R$631) ser maior que a mediana (R$502) aponta para uma distribuicao assimetrica a direita. Isso acontece porque existem poucos estados com renda muito alta — como o Distrito Federal — que puxam a media para cima. A media e uma medida NAO RESISTENTE a valores extremos: ela muda bastante com outliers. Ja a mediana e RESISTENTE, por isso permanece mais proxima da maioria dos dados. Isso indica que a maioria dos estados tem renda abaixo da media, e poucos concentram rendas muito mais elevadas."

### Media < Mediana → assimetria a esquerda

**Resposta pronta:**
"A media menor que a mediana indica distribuicao assimetrica a esquerda. Existem poucos valores muito baixos que puxam a media para baixo, enquanto a mediana se mantem mais estavel por ser resistente a extremos."

### Por que usar mediana e nao media para renda?

**Resposta pronta:**
"A mediana e mais adequada para renda porque e RESISTENTE a valores extremos. Como a distribuicao de renda e assimetrica a direita, a media fica inflada pelos poucos com renda muito alta e nao representa bem a maioria da populacao. A mediana reflete o valor do centro real da distribuicao."

### Como descrever Pearson?

**Tabela rapida:**

| Valor | Força |
|---|---|
| 0.9 a 1.0 / -0.9 a -1.0 | Muito forte |
| 0.7 a 0.9 / -0.7 a -0.9 | Forte |
| 0.5 a 0.7 / -0.5 a -0.7 | Moderada |
| 0 a 0.5 / 0 a -0.5 | Fraca |

**Resposta pronta:**
"O coeficiente de Pearson de [valor] indica correlacao linear [positiva/negativa] e [fraca/moderada/forte] entre [X] e [Y]. Ou seja, em geral, quanto maior [X], [maior/menor] tende a ser [Y]. Sem o outlier [quem], o Pearson passa a [valor], mostrando relacao ainda [mais forte/mais fraca]."

### Como descrever o histograma?

| O que voce ve | Como escrever |
|---|---|
| Cauda longa a direita, Media > Mediana | "assimetrica a direita" |
| Cauda longa a esquerda, Media < Mediana | "assimetrica a esquerda" |
| Media aprox = Mediana, forma de sino | "aproximadamente simetrica" |
| Dois picos | "bimodal" |
