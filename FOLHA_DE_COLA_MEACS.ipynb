## 3. BIVARIADA CATEGORICA — TABELA CRUZADA + BARRAS 100%

Usar quando: DUAS variaveis categoricas (ex: cor/raca x acesso internet)

TROCA: COL1 = variavel das linhas (a principal), COL2 = a que divide em grupos

```python
# TABELA CRUZADA — frequencia absoluta e relativa
COL1 = 'TP_COR_RACA'  # <-- MUDA AQUI
COL2 = 'Q025'          # <-- MUDA AQUI

print('Frequencia absoluta:')
print(pd.crosstab(df[COL1], df[COL2]))
print()
print('Frequencia relativa (% por linha):')
print((pd.crosstab(df[COL1], df[COL2], normalize='index') * 100).round(2))
```

```python
# BARRAS 100% NORMALIZADAS (distribuicao condicional)
COL1 = 'TP_COR_RACA'  # <-- MUDA AQUI
COL2 = 'Q025'          # <-- MUDA AQUI

pivot = df.groupby([COL1, COL2]).size().reset_index(name='count')
pivot['total'] = pivot.groupby(COL1)['count'].transform('sum')
pivot['pct'] = pivot['count'] / pivot['total']

alt.Chart(pivot).mark_bar().encode(
    x=alt.X(COL1 + ':N', title=COL1),
    y=alt.Y('pct:Q', title='Proporcao', axis=alt.Axis(format='%')),
    color=alt.Color(COL2 + ':N'),
    tooltip=[COL1 + ':N', COL2 + ':N', 'pct:Q']
).properties(title='Distribuicao condicional de ' + COL2 + ' por ' + COL1, width=500)
```

```python
# RESUMO DOS 5 NUMEROS + MEDIA + DESVIO PADRAO + AIQ + LIMITES OUTLIER
COLUNA = 'renda_per_capita'  # <-- MUDA AQUI

Q1 = df[COLUNA].quantile(0.25)
Q3 = df[COLUNA].quantile(0.75)
AIQ = Q3 - Q1

print(f'Minimo:           {df[COLUNA].min():.2f}')
print(f'Q1 (25%):         {Q1:.2f}')
print(f'Mediana (50%):    {df[COLUNA].median():.2f}')
print(f'Media:            {df[COLUNA].mean():.2f}')
print(f'Q3 (75%):         {Q3:.2f}')
print(f'Maximo:           {df[COLUNA].max():.2f}')
print(f'Desvio Padrao:    {df[COLUNA].std():.2f}')
print(f'AIQ (Q3-Q1):      {AIQ:.2f}')
print(f'Limite inf (1.5): {Q1 - 1.5*AIQ:.2f}  <- abaixo disso = outlier')
print(f'Limite sup (1.5): {Q3 + 1.5*AIQ:.2f}  <- acima disso = outlier')
```

## 5. DETECTAR OUTLIERS — REGRA 1.5 x AIQ

Formula:
- Limite inferior = Q1 - 1.5 x AIQ
- Limite superior = Q3 + 1.5 x AIQ
- FORA desses limites = outlier

```python
# DETECTAR QUEM SAO OS OUTLIERS
COLUNA = 'renda_per_capita'  # <-- MUDA AQUI
COL_NOME = 'estado'          # <-- coluna que identifica cada linha (nome do municipio, estado...)

Q1 = df[COLUNA].quantile(0.25)
Q3 = df[COLUNA].quantile(0.75)
AIQ = Q3 - Q1
lim_inf = Q1 - 1.5 * AIQ
lim_sup = Q3 + 1.5 * AIQ

outliers = df[(df[COLUNA] < lim_inf) | (df[COLUNA] > lim_sup)]
print(f'Limite inferior: {lim_inf:.2f}')
print(f'Limite superior: {lim_sup:.2f}')
print(f'Outliers encontrados: {len(outliers)}')
print(outliers[[COL_NOME, COLUNA]])
```

INTERPRETACAO MODELO:
"Pela regra 1.5 x AIQ, o limite superior e [valor]. O(s) dado(s) [quem] ultrapassa(m) esse limite com [valor], sendo considerado(s) outlier(s). Esse valor atipico [influencia/nao influencia muito] a media, puxando-a para [cima/baixo]."

## 9. RESPOSTAS TEORICAS — ADAPTE OS VALORES

Essas perguntas aparecem sempre. So troca o valor e a variavel.

### Media > Mediana — o que significa?

Resposta pronta:
"O fato de a media ([valor]) ser maior que a mediana ([valor]) aponta para uma distribuicao assimetrica a direita. Isso acontece porque existem poucos valores muito altos que puxam a media para cima — ja que ela e uma medida NAO RESISTENTE a valores extremos. A mediana, por ser resistente, nao sofre essa influencia. Isso indica que a maioria dos casos tem valores baixos e poucos casos tem valores muito mais elevados, gerando desigualdade na distribuicao."

### Media < Mediana — o que significa?

Resposta pronta:
"O fato de a media ([valor]) ser menor que a mediana ([valor]) aponta para uma distribuicao assimetrica a esquerda. Existem poucos valores muito baixos que arrastam a media para baixo, enquanto a mediana permanece mais estavel por ser uma medida resistente a valores extremos."

### Por que usar mediana em vez de media (para renda)?

Resposta pronta:
"Para variaveis de renda, a mediana e mais adequada porque e uma medida RESISTENTE — nao e influenciada por valores extremos. Como a distribuicao de renda e assimetrica a direita, a media fica inflada pelos poucos com renda muito alta e nao representa bem a maioria da populacao. A mediana representa o valor central real da distribuicao."

### O que e um outlier pela regra 1.5xAIQ?

Resposta pronta:
"Um valor e considerado outlier quando esta abaixo de Q1 - 1.5 x AIQ ou acima de Q3 + 1.5 x AIQ. No caso de [variavel], o limite superior e [valor]. O dado [quem] ultrapassa esse limite com [valor], sendo portanto um outlier segundo essa regra."

### Como descrever Correlacao de Pearson?

Resposta pronta:
"O coeficiente de Pearson calculado foi de [valor], indicando uma correlacao linear [positiva/negativa] e [fraca/moderada/forte] entre [X] e [Y]. Isso significa que, em geral, quanto maior [X], [maior/menor] tende a ser [Y]. Ao remover o outlier [quem], o coeficiente passa a [valor], mostrando que a relacao e [mais forte/mais fraca] sem esse ponto atipico."
