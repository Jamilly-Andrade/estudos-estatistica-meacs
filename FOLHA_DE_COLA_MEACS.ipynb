## 1. VARIAVEL CATEGORICA — GRAFICO DE BARRAS

## 3. BIVARIADA CATEGORICA — TABELA CRUZADA + BARRAS 100%

Usar quando: DUAS variaveis categoricas (ex: cor/raca x acesso internet)

TROCA: COL1 = variavel das linhas (a principal), COL2 = a que divide em grupos

```python
# TABELA CRUZADA — frequencia absoluta e relativa
COL1 = 'TP_COR_RACA'  # <-- MUDA AQUI
COL2 = 'Q025'          # <-- MUDA AQUI

print('Frequencia absoluta:')
print(pd.crosstab(df[COL1], df[COL2]))
print()
print('Frequencia relativa (% por linha):')
print((pd.crosstab(df[COL1], df[COL2], normalize='index') * 100).round(2))
```

```python
# BARRAS 100% NORMALIZADAS (distribuicao condicional)
COL1 = 'TP_COR_RACA'  # <-- MUDA AQUI
COL2 = 'Q025'          # <-- MUDA AQUI

pivot = df.groupby([COL1, COL2]).size().reset_index(name='count')
pivot['total'] = pivot.groupby(COL1)['count'].transform('sum')
pivot['pct'] = pivot['count'] / pivot['total']

alt.Chart(pivot).mark_bar().encode(
    x=alt.X(COL1 + ':N', title=COL1),
    y=alt.Y('pct:Q', title='Proporcao', axis=alt.Axis(format='%')),
    color=alt.Color(COL2 + ':N'),
    tooltip=[COL1 + ':N', COL2 + ':N', 'pct:Q']
).properties(title='Distribuicao condicional de ' + COL2 + ' por ' + COL1, width=500)
```

INTERPRETACAO MODELO:
"Ao analisar a distribuicao condicional de [COL2] em relacao a [COL1], observamos que [grupo X] tem o maior percentual de [categoria Y] com [Z]%, enquanto [grupo W] tem o menor com [V]%. Isso sugere que existe associacao entre as duas variaveis."

---
## 4. VARIAVEL NUMERICA — HISTOGRAMA + RESUMO

Usar quando: variavel numerica (renda, nota, temperatura, analfabetismo...)

TROCA: COLUNA pelo nome da coluna numerica

```python
# HISTOGRAMA
COLUNA = 'renda_per_capita'  # <-- MUDA AQUI

alt.Chart(df).mark_bar(color='steelblue', opacity=0.85).encode(
    x=alt.X(COLUNA + ':Q', bin=alt.Bin(maxbins=30), title=COLUNA),
    y=alt.Y('count():Q', title='Frequencia')
).properties(title='Histograma de ' + COLUNA, width=500)
```

```python
# RESUMO DOS 5 NUMEROS + MEDIA + DESVIO PADRAO + AIQ + LIMITES OUTLIER
COLUNA = 'renda_per_capita'  # <-- MUDA AQUI

Q1 = df[COLUNA].quantile(0.25)
Q3 = df[COLUNA].quantile(0.75)
AIQ = Q3 - Q1

print(f'Minimo:           {df[COLUNA].min():.2f}')
print(f'Q1 (25%):         {Q1:.2f}')
print(f'Mediana (50%):    {df[COLUNA].median():.2f}')
print(f'Media:            {df[COLUNA].mean():.2f}')
print(f'Q3 (75%):         {Q3:.2f}')
print(f'Maximo:           {df[COLUNA].max():.2f}')
print(f'Desvio Padrao:    {df[COLUNA].std():.2f}')
print(f'AIQ (Q3-Q1):      {AIQ:.2f}')
print(f'Limite inf (1.5): {Q1 - 1.5*AIQ:.2f}  <- abaixo disso = outlier')
print(f'Limite sup (1.5): {Q3 + 1.5*AIQ:.2f}  <- acima disso = outlier')
```

COMO DESCREVER A FORMA DO HISTOGRAMA:

| O que voce ve | Como escrever |
|---|---|
| Cauda longa para direita, Media > Mediana | "assimetrica a direita" |
| Cauda longa para esquerda, Media < Mediana | "assimetrica a esquerda" |
| Media aprox. = Mediana, forma de sino | "aproximadamente simetrica" |
| Dois picos | "bimodal" |

INTERPRETACAO MODELO:
"A distribuicao de [COLUNA] e assimetrica a direita, com a maioria dos valores concentrados em [faixa]. A media ([X]) e maior que a mediana ([Y]), confirmando a assimetria — poucos valores altos puxam a media. O desvio padrao de [Z] indica [alta/baixa] dispersao."

## 5. DETECTAR OUTLIERS — REGRA 1.5 x AIQ

Formula:
- Limite inferior = Q1 - 1.5 x AIQ
- Limite superior = Q3 + 1.5 x AIQ
- FORA desses limites = outlier

```python
# DETECTAR QUEM SAO OS OUTLIERS
COLUNA = 'renda_per_capita'  # <-- MUDA AQUI
COL_NOME = 'estado'          # <-- coluna que identifica cada linha (nome do municipio, estado...)

Q1 = df[COLUNA].quantile(0.25)
Q3 = df[COLUNA].quantile(0.75)
AIQ = Q3 - Q1
lim_inf = Q1 - 1.5 * AIQ
lim_sup = Q3 + 1.5 * AIQ

outliers = df[(df[COLUNA] < lim_inf) | (df[COLUNA] > lim_sup)]
print(f'Limite inferior: {lim_inf:.2f}')
print(f'Limite superior: {lim_sup:.2f}')
print(f'Outliers encontrados: {len(outliers)}')
print(outliers[[COL_NOME, COLUNA]])
```

INTERPRETACAO MODELO:
"Pela regra 1.5 x AIQ, o limite superior e [valor]. O(s) dado(s) [quem] ultrapassa(m) esse limite com [valor], sendo considerado(s) outlier(s). Esse valor atipico [influencia/nao influencia muito] a media, puxando-a para [cima/baixo]."

## 6. BIVARIADA NUMERICA — DISPERSAO + PEARSON

Usar quando: DUAS variaveis numericas (ex: renda x analfabetismo, nota x renda)

TROCA: COL_X = variavel explicativa (eixo X), COL_Y = variavel resposta (eixo Y)

```python
# GRAFICO DE DISPERSAO
COL_X = 'tx_analfabetismo'  # <-- variavel explicativa (causa)
COL_Y = 'renda_per_capita'  # <-- variavel resposta (efeito)
COL_LABEL = 'estado'        # <-- coluna com nome/rotulo (aparece no tooltip)

alt.Chart(df).mark_point(filled=True, size=80, color='steelblue').encode(
    x=alt.X(COL_X + ':Q', title=COL_X),
    y=alt.Y(COL_Y + ':Q', title=COL_Y),
    tooltip=[COL_LABEL + ':N', COL_X + ':Q', COL_Y + ':Q']
).properties(title=COL_Y + ' x ' + COL_X, width=500)
```

```python
# CORRELACAO DE PEARSON — com todos os dados
COL_X = 'tx_analfabetismo'  # <-- MUDA AQUI
COL_Y = 'renda_per_capita'  # <-- MUDA AQUI

pearson = df[COL_X].corr(df[COL_Y])
print(f'Correlacao de Pearson entre {COL_X} e {COL_Y}: {pearson:.4f}')
```

```python
# CORRELACAO SEM O OUTLIER (ex: sem o Distrito Federal)
COL_X = 'tx_analfabetismo'      # <-- MUDA AQUI
COL_Y = 'renda_per_capita'      # <-- MUDA AQUI
COL_OUTLIER = 'estado'           # <-- coluna onde esta o outlier
VALOR_OUTLIER = 'Distrito Federal'  # <-- valor a remover

df_sem = df[df[COL_OUTLIER] != VALOR_OUTLIER]
print(f'Com outlier:   {df[COL_X].corr(df[COL_Y]):.4f}')
print(f'Sem outlier:   {df_sem[COL_X].corr(df_sem[COL_Y]):.4f}')
```

## 7. SERIE TEMPORAL

Usar quando: dados ao longo do tempo (temperatura, chuva, custo, taxa...)

TROCA: COL_DATA = coluna de datas, COL_VALOR = coluna numerica a analisar

```python
# SERIE TEMPORAL — linha simples
COL_DATA = 'DATA'       # <-- MUDA AQUI
COL_VALOR = 'temp_avg'  # <-- MUDA AQUI

df[COL_DATA] = pd.to_datetime(df[COL_DATA])  # converte para data

alt.Chart(df).mark_line(color='steelblue').encode(
    x=alt.X(COL_DATA + ':T', title='Data'),
    y=alt.Y(COL_VALOR + ':Q', title=COL_VALOR),
    tooltip=[COL_DATA + ':T', COL_VALOR + ':Q']
).properties(title='Serie temporal de ' + COL_VALOR, width=600)
```

```python
# SERIE TEMPORAL AGREGADA POR MES (media mensal)
COL_DATA = 'DATA'       # <-- MUDA AQUI
COL_VALOR = 'temp_avg'  # <-- MUDA AQUI

df[COL_DATA] = pd.to_datetime(df[COL_DATA])
df_mensal = (
    df.set_index(COL_DATA)[COL_VALOR]
    .resample('M')
    .mean()
    .reset_index()
)
df_mensal.columns = ['data_mes', 'media_mensal']

alt.Chart(df_mensal).mark_line(color='coral').encode(
    x=alt.X('data_mes:T', title='Mes'),
    y=alt.Y('media_mensal:Q', title='Media mensal de ' + COL_VALOR),
    tooltip=['data_mes:T', 'media_mensal:Q']
).properties(title='Media mensal de ' + COL_VALOR, width=600)
```

INTERPRETACAO MODELO:
"O grafico de serie temporal de [COL_VALOR] mostra um comportamento [ciclico/crescente/decrescente/estavel] ao longo do tempo. Os picos ocorrem em [periodo] e os vales em [periodo]. Isso sugere [sazonalidade anual / tendencia de aumento / anomalia em certo periodo]."

## 8. BOXPLOT COMPARATIVO (lado a lado)

Usar quando: comparar distribuicao numerica entre grupos (homens vs mulheres, brancos vs negros, regioes...)

DOIS casos: coluna de grupo ja existe OU colunas separadas por grupo (usa melt)

```python
# CASO 1: ja tem coluna de grupo na base
COL_NUM = 'renda_per_capita'  # <-- variavel numerica
COL_GRUPO = 'sexo'             # <-- coluna de grupo (categorica)

alt.Chart(df).mark_boxplot(extent=1.5).encode(
    x=alt.X(COL_GRUPO + ':N', title=COL_GRUPO),
    y=alt.Y(COL_NUM + ':Q', title=COL_NUM),
    color=alt.Color(COL_GRUPO + ':N')
).properties(title='Boxplot de ' + COL_NUM + ' por ' + COL_GRUPO, width=400)
```

```python
# CASO 2: colunas separadas por grupo — usa .melt() para juntar
# Ex: base tem coluna "renda_homem" e "renda_mulher" separadas
COL_ID = 'estado'                           # <-- coluna que identifica cada linha
COLS_VARS = ['renda_homem', 'renda_mulher']  # <-- colunas que vao virar grupos

df_long = df.melt(
    id_vars=[COL_ID],
    value_vars=COLS_VARS,
    var_name='grupo',
    value_name='valor'
)

alt.Chart(df_long).mark_boxplot(extent=1.5).encode(
    x=alt.X('grupo:N', title='Grupo'),
    y=alt.Y('valor:Q', title='Valor'),
    color=alt.Color('grupo:N')
).properties(title='Boxplot comparativo', width=400)
```

O QUE OLHAR NO BOXPLOT:

| Parte de baixo da caixa | Q1 (25%) |

| Bigodes (tracos) | Min e Max DENTRO do 1.5xAIQ |
| Pontinhos fora dos bigodes | OUTLIERS |
| Caixa mais alta = valores maiores | Medianas mais altas |

INTERPRETACAO MODELO:
"Os boxplots comparativos mostram que [grupo A] tem mediana de [X] e [grupo B] tem mediana de [Y]. Portanto, [grupo A/B] tende a ter valores mais altos. A distribuicao de [grupo A] e assimetrica a [direita/esquerda] pois o bigode [superior/inferior] e mais longo. O Distrito Federal aparece como outlier em ambos os grupos pela regra 1.5xAIQ, com valor de [Z]."

## 9. RESPOSTAS TEORICAS — ADAPTE OS VALORES

Essas perguntas aparecem sempre. So troca o valor e a variavel.

### Media > Mediana — o que significa?

Resposta pronta:
"O fato de a media ([valor]) ser maior que a mediana ([valor]) aponta para uma distribuicao assimetrica a direita. Isso acontece porque existem poucos valores muito altos que puxam a media para cima — ja que ela e uma medida NAO RESISTENTE a valores extremos. A mediana, por ser resistente, nao sofre essa influencia. Isso indica que a maioria dos casos tem valores baixos e poucos casos tem valores muito mais elevados, gerando desigualdade na distribuicao."

### Media < Mediana — o que significa?

Resposta pronta:
"O fato de a media ([valor]) ser menor que a mediana ([valor]) aponta para uma distribuicao assimetrica a esquerda. Existem poucos valores muito baixos que arrastam a media para baixo, enquanto a mediana permanece mais estavel por ser uma medida resistente a valores extremos."

### Por que usar mediana em vez de media (para renda)?

Resposta pronta:
"Para variaveis de renda, a mediana e mais adequada porque e uma medida RESISTENTE — nao e influenciada por valores extremos. Como a distribuicao de renda e assimetrica a direita, a media fica inflada pelos poucos com renda muito alta e nao representa bem a maioria da populacao. A mediana representa o valor central real da distribuicao."

### O que e um outlier pela regra 1.5xAIQ?

Resposta pronta:
"Um valor e considerado outlier quando esta abaixo de Q1 - 1.5 x AIQ ou acima de Q3 + 1.5 x AIQ. No caso de [variavel], o limite superior e [valor]. O dado [quem] ultrapassa esse limite com [valor], sendo portanto um outlier segundo essa regra."

### Como descrever Correlacao de Pearson?

Resposta pronta:
"O coeficiente de Pearson calculado foi de [valor], indicando uma correlacao linear [positiva/negativa] e [fraca/moderada/forte] entre [X] e [Y]. Isso significa que, em geral, quanto maior [X], [maior/menor] tende a ser [Y]. Ao remover o outlier [quem], o coeficiente passa a [valor], mostrando que a relacao e [mais forte/mais fraca] sem esse ponto atipico."
